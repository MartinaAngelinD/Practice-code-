```python
# ============================================================
# CALCULATING COSINE SIMILARITY MATHEMATICALLY IN PYTHON
# ============================================================

# ------------------------------------------------------------
# 1. DEFINE TWO VECTORS
# ------------------------------------------------------------

vector_A = [2, 4, 1]
vector_B = [1, 3, 5]

print("Vector A:", vector_A)
print("Vector B:", vector_B)


# ------------------------------------------------------------
# 2. DOT PRODUCT - FROM SCRATCH
# ------------------------------------------------------------

def dot_product(A, B):
    total = 0
    for a, b in zip(A, B):
        total += a * b
    return total

dp = dot_product(vector_A, vector_B)
print("\nDot Product (A . B):", dp)


# ------------------------------------------------------------
# 3. MAGNITUDE - FROM SCRATCH
# ------------------------------------------------------------

import math

def magnitude(A):
    total = 0
    for a in A:
        total += a ** 2
    return math.sqrt(total)

mag_A = magnitude(vector_A)
mag_B = magnitude(vector_B)

print("\nMagnitude of A:", round(mag_A, 4))
print("Magnitude of B:", round(mag_B, 4))


# ------------------------------------------------------------
# 4. COSINE SIMILARITY - FROM SCRATCH
# ------------------------------------------------------------

def cosine_similarity_manual(A, B):
    dp = dot_product(A, B)
    mag_A = magnitude(A)
    mag_B = magnitude(B)
    similarity = dp / (mag_A * mag_B)
    return similarity

result_manual = cosine_similarity_manual(vector_A, vector_B)
print("\nCosine Similarity (Manual):", round(result_manual, 4))


# ------------------------------------------------------------
# 5. VERIFY USING NUMPY
# ------------------------------------------------------------

import numpy as np

A = np.array(vector_A)
B = np.array(vector_B)

dot = np.dot(A, B)
norm_A = np.linalg.norm(A)
norm_B = np.linalg.norm(B)

result_numpy = dot / (norm_A * norm_B)
print("Cosine Similarity (NumPy):", round(result_numpy, 4))


# ------------------------------------------------------------
# 6. VERIFY USING SCIKIT-LEARN
# ------------------------------------------------------------

!pip install -q scikit-learn

from sklearn.metrics.pairwise import cosine_similarity

# Convert vectors into 2D arrays
A_2d = np.array(vector_A).reshape(1, -1)
B_2d = np.array(vector_B).reshape(1, -1)

result_sklearn = cosine_similarity(A_2d, B_2d)[0][0]
print("Cosine Similarity (scikit-learn):", round(result_sklearn, 4))


# ------------------------------------------------------------
# 7. COMPARE ALL THREE RESULTS
# ------------------------------------------------------------

print("\n----------------------------------------")
print("COMPARISON")
print("----------------------------------------")
print("Manual      :", round(result_manual, 6))
print("NumPy       :", round(result_numpy, 6))
print("scikit-learn:", round(result_sklearn, 6))

# Check whether all results are equal
assert round(result_manual, 4) == round(result_numpy, 4) == round(result_sklearn, 4)
print("\nAll three methods match!")


# ------------------------------------------------------------
# 8. TEST MORE EXAMPLES
# ------------------------------------------------------------

examples = {
    "Identical vectors": ([1, 2, 3], [1, 2, 3]),
    "Opposite vectors": ([1, 2, 3], [-1, -2, -3]),
    "Perpendicular vectors": ([1, 0], [0, 1]),
    "Scaled vectors": ([1, 2, 3], [2, 4, 6])
}

print("\n----------------------------------------")
print("MORE EXAMPLES")
print("----------------------------------------")
for name, (a, b) in examples.items():
    sim = cosine_similarity_manual(a, b)
    print(f"{name:30s} -> {round(sim, 4)}")


# ------------------------------------------------------------
# 9. SIMPLE INTERPRETATION
# ------------------------------------------------------------

print("\n----------------------------------------")
print("INTERPRETATION")
print("----------------------------------------")
print("Identical vectors       -> 1")
print("Opposite vectors        -> -1")
print("Perpendicular vectors   -> 0")
print("Same direction vectors  -> 1")
```
