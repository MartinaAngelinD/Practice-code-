```python
# ============================================================
# CRUD OPERATIONS IN CHROMADB
# Google Colab - Free CPU
# ============================================================

# 1. INSTALL CHROMADB
!pip install -q chromadb


# 2. IMPORT CHROMADB
import chromadb


# 3. CREATE CLIENT
# In-memory client: data is cleared when the Colab runtime restarts
client = chromadb.Client()


# 4. CREATE COLLECTION
# Delete old collection if it already exists
try:
    client.delete_collection("crud_demo")
except Exception:
    pass

collection = client.create_collection(name="crud_demo")

print("Collection created:", collection.name)
print("Current document count:", collection.count())


# ============================================================
# 5. CREATE - ADD DOCUMENTS
# ============================================================

collection.add(
    documents=[
        "Python is a popular programming language for AI.",
        "The weather today is sunny and warm.",
        "Machine learning models learn patterns from data.",
        "I enjoy playing football on weekends.",
        "Deep learning uses neural networks with many layers."
    ],
    metadatas=[
        {"category": "tech"},
        {"category": "weather"},
        {"category": "tech"},
        {"category": "sports"},
        {"category": "tech"}
    ],
    ids=[
        "doc1",
        "doc2",
        "doc3",
        "doc4",
        "doc5"
    ]
)

print("\nDocuments added!")
print("Total document count:", collection.count())


# ============================================================
# 6. READ - PEEK
# ============================================================

print("\n--- PEEK: FIRST 3 DOCUMENTS ---")

preview = collection.peek(limit=3)

for doc_id, doc in zip(preview["ids"], preview["documents"]):
    print(doc_id, "->", doc)


# ============================================================
# 7. READ - GET DOCUMENTS BY ID
# ============================================================

print("\n--- GET DOCUMENTS BY ID ---")

result = collection.get(ids=["doc1", "doc3"])

for doc_id, doc, meta in zip(result["ids"], result["documents"], result["metadatas"]):
    print(f"{doc_id} | {meta} | {doc}")


# ============================================================
# 8. READ - SEMANTIC SEARCH
# ============================================================

print("\n--- SEMANTIC SEARCH ---")

results = collection.query(
    query_texts=["Tell me about AI and neural networks"],
    n_results=2
)

for doc_id, doc, distance in zip(results["ids"][0], results["documents"][0], results["distances"][0]):
    print(f"{doc_id} | distance={distance:.4f} | {doc}")


# ============================================================
# 9. READ - FILTER USING METADATA
# ============================================================

print("\n--- FILTERED SEARCH: SPORTS ---")

results = collection.query(
    query_texts=["something fun to do outside"],
    n_results=3,
    where={"category": "sports"}
)

for doc_id, doc, distance in zip(results["ids"][0], results["documents"][0], results["distances"][0]):
    print(f"{doc_id} | distance={distance:.4f} | {doc}")


# ============================================================
# 10. UPDATE - MODIFY EXISTING DOCUMENT
# ============================================================

print("\n--- UPDATE DOCUMENT ---")

collection.update(
    ids=["doc4"],
    documents=["I enjoy playing football and basketball on weekends."],
    metadatas=[{
        "category": "sports",
        "updated": True
    }]
)

updated = collection.get(ids=["doc4"])

print("Updated document:")
print(updated["documents"])

print("Updated metadata:")
print(updated["metadatas"])


# ============================================================
# 11. UPSERT - ADD OR UPDATE
# ============================================================

print("\n--- UPSERT ---")

collection.upsert(
    documents=[
        "Deep learning uses neural networks with many layers and GPUs.",
        "The Eiffel Tower is located in Paris."
    ],
    ids=["doc5", "doc6"]
)

print("Total count after upsert:", collection.count())

upserted = collection.get(ids=["doc5", "doc6"])

print("Documents after upsert:")
for doc_id, doc in zip(upserted["ids"], upserted["documents"]):
    print(doc_id, "->", doc)


# ============================================================
# 12. DELETE - REMOVE DOCUMENT BY ID
# ============================================================

print("\n--- DELETE BY ID ---")

collection.delete(ids=["doc2"])

print("Total count after deleting doc2:", collection.count())
print("Remaining IDs:")
print(collection.get()["ids"])


# ============================================================
# 13. DELETE - REMOVE DOCUMENTS BY FILTER
# ============================================================

print("\n--- DELETE BY CATEGORY ---")

collection.delete(where={"category": "tech"})

print("Total count after deleting tech documents:", collection.count())
print("Remaining IDs:")
print(collection.get()["ids"])


# ============================================================
# 14. FINAL STATE
# ============================================================

print("\n--- FINAL COLLECTION STATE ---")

final_state = collection.get()

print("Final document count:", collection.count())
for doc_id, doc, meta in zip(final_state["ids"], final_state["documents"], final_state["metadatas"]):
    print(f"{doc_id} | {meta} | {doc}")


# ============================================================
# 15. CRUD SUMMARY
# ============================================================

print("\n========================================")
print("             CRUD SUMMARY")
print("========================================")
print("CREATE  -> collection.add()")
print("READ    -> collection.get()")
print("READ    -> collection.query()")
print("READ    -> collection.peek()")
print("UPDATE  -> collection.update()")
print("UPSERT  -> collection.upsert()")
print("DELETE  -> collection.delete()")
print("========================================")
```
