```python
# ============================================================
# INTRODUCTION TO VECTOR DATABASES USING CHROMADB
# ============================================================

# ------------------------------------------------------------
# 1. INSTALL REQUIRED LIBRARIES
# ------------------------------------------------------------

!pip install -q chromadb sentence-transformers


# ------------------------------------------------------------
# 2. IMPORT CHROMADB
# ------------------------------------------------------------

import chromadb

print("ChromaDB imported successfully!")


# ------------------------------------------------------------
# 3. CREATE CHROMADB CLIENT
# ------------------------------------------------------------

# In-memory client
# Data will be lost when the Colab runtime restarts

client = chromadb.Client()

print("ChromaDB client created successfully!")


# ------------------------------------------------------------
# 4. CREATE A COLLECTION
# ------------------------------------------------------------

# Delete collection if it already exists
try:
    client.delete_collection("my_first_collection")
except Exception:
    pass

collection = client.create_collection(
    name="my_first_collection"
)

print("Collection created:", collection.name)


# ------------------------------------------------------------
# 5. ADD DOCUMENTS
# ------------------------------------------------------------

collection.add(
    documents=[
        "Python is a popular programming language for AI.",
        "The weather today is sunny and warm.",
        "Machine learning models learn patterns from data.",
        "I enjoy playing football on weekends.",
        "Deep learning uses neural networks with many layers.",
        "Paris is the capital city of France."
    ],
    ids=[
        "doc1",
        "doc2",
        "doc3",
        "doc4",
        "doc5",
        "doc6"
    ]
)

print("Documents added!")
print("Total documents:", collection.count())


# ------------------------------------------------------------
# 6. QUERY BY MEANING - SEMANTIC SEARCH
# ------------------------------------------------------------

print("\n========================================")
print("SEMANTIC SEARCH - QUERY 1")
print("========================================")

results = collection.query(
    query_texts=["Tell me about AI and neural networks"],
    n_results=2
)

for doc, distance in zip(
    results["documents"][0],
    results["distances"][0]
):
    print(f"Distance: {distance:.4f} | {doc}")


# ------------------------------------------------------------
# 7. ANOTHER SEMANTIC SEARCH
# ------------------------------------------------------------

print("\n========================================")
print("SEMANTIC SEARCH - QUERY 2")
print("========================================")

results = collection.query(
    query_texts=["What is the capital of France?"],
    n_results=2
)

for doc, distance in zip(
    results["documents"][0],
    results["distances"][0]
):
    print(f"Distance: {distance:.4f} | {doc}")


# ------------------------------------------------------------
# 8. ADD DOCUMENTS WITH METADATA
# ------------------------------------------------------------

collection.add(
    documents=[
        "The new smartphone has an amazing camera.",
        "Stock prices rose sharply after the earnings report."
    ],
    metadatas=[
        {"category": "technology"},
        {"category": "finance"}
    ],
    ids=[
        "doc7",
        "doc8"
    ]
)

print("\nDocuments with metadata added!")


# ------------------------------------------------------------
# 9. QUERY USING METADATA FILTER
# ------------------------------------------------------------

print("\n========================================")
print("METADATA FILTER SEARCH")
print("========================================")

results = collection.query(
    query_texts=["Tell me about new gadgets"],
    n_results=2,
    where={"category": "technology"}
)

for doc, distance in zip(
    results["documents"][0],
    results["distances"][0]
):
    print(f"Distance: {distance:.4f} | {doc}")


# ------------------------------------------------------------
# 10. USE HUGGING FACE EMBEDDINGS
# ------------------------------------------------------------

from chromadb.utils import embedding_functions

# Create Hugging Face embedding function
hf_embedder = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)

print("\nHugging Face embedding model loaded!")


# ------------------------------------------------------------
# 11. CREATE CUSTOM EMBEDDING COLLECTION
# ------------------------------------------------------------

# Delete collection if it already exists
try:
    client.delete_collection("custom_embedding_collection")
except Exception:
    pass

custom_collection = client.create_collection(
    name="custom_embedding_collection",
    embedding_function=hf_embedder
)

print("Custom embedding collection created!")


# ------------------------------------------------------------
# 12. ADD DOCUMENTS TO CUSTOM COLLECTION
# ------------------------------------------------------------

custom_collection.add(
    documents=[
        "I love dogs",
        "I like puppies",
        "The stock market crashed today"
    ],
    ids=[
        "a1",
        "a2",
        "a3"
    ]
)

print("Documents added to custom collection!")


# ------------------------------------------------------------
# 13. SEARCH USING HUGGING FACE EMBEDDINGS
# ------------------------------------------------------------

print("\n========================================")
print("HUGGING FACE SEMANTIC SEARCH")
print("========================================")

results = custom_collection.query(
    query_texts=["puppy lovers"],
    n_results=2
)

for doc, distance in zip(
    results["documents"][0],
    results["distances"][0]
):
    print(f"Distance: {distance:.4f} | {doc}")


# ------------------------------------------------------------
# 14. UPDATE A DOCUMENT
# ------------------------------------------------------------

collection.update(
    ids=["doc4"],
    documents=[
        "I enjoy playing football and basketball on weekends."
    ]
)

print("\nUpdated doc4")


# ------------------------------------------------------------
# 15. DELETE A DOCUMENT
# ------------------------------------------------------------

collection.delete(
    ids=["doc2"]
)

print("Deleted doc2")

print("Remaining documents:", collection.count())


# ------------------------------------------------------------
# 16. PERSISTENT STORAGE
# ------------------------------------------------------------

# Create a persistent ChromaDB client
persistent_client = chromadb.PersistentClient(
    path="/content/chroma_db"
)

# Create or get a persistent collection
persistent_collection = persistent_client.get_or_create_collection(
    name="persistent_demo"
)


# ------------------------------------------------------------
# 17. ADD DATA TO PERSISTENT DATABASE
# ------------------------------------------------------------

# Check if p1 already exists before adding
existing = persistent_collection.get(
    ids=["p1"]
)

if not existing["ids"]:
    persistent_collection.add(
        documents=[
            "This data will be saved to disk."
        ],
        ids=["p1"]
    )

print("\n========================================")
print("PERSISTENT STORAGE")
print("========================================")

print("Saved to disk at /content/chroma_db")
print("Count:", persistent_collection.count())


# ------------------------------------------------------------
# 18. FINAL SUMMARY
# ------------------------------------------------------------

print("\n========================================")
print("PROGRAM COMPLETED SUCCESSFULLY")
print("========================================")

print("""
Concepts demonstrated:
1. ChromaDB installation
2. ChromaDB client creation
3. Collection creation
4. Adding documents
5. Semantic search
6. Metadata filtering
7. Hugging Face embeddings
8. Updating documents
9. Deleting documents
10. Persistent storage
""")
```
