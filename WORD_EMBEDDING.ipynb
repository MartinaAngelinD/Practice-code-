```python
import sys
# ============================================================
# WORD EMBEDDING USING ONE-HOT ENCODING AND WORD2VEC
# ============================================================

# -------------------
# PART 1: ONE-HOT ENCODING
# -------------------

from sklearn.preprocessing import OneHotEncoder

# Define the words
words = [
    ['cat'],
    ['dog'],
    ['apple'],
    ['car']
]

# Create One-Hot Encoder
encoder = OneHotEncoder(sparse_output=False)

# Convert words into One-Hot vectors
result = encoder.fit_transform(words)

# Display words
print("Words:", encoder.categories_[0])

# Display One-Hot Embeddings
print("\nOne-Hot Embeddings:")
print(result)

# -------------------
# PART 2: WORD2VEC
# -------------------

# Install gensim
!pip install -q gensim

# Import Word2Vec
from gensim.models import Word2Vec

# Training sentences
sentences = [
    ["python", "is", "a", "language"],
    ["java", "is", "a", "language"],
    ["python", "is", "easy"],
    ["java", "is", "easy"],
    ["python", "is", "powerful"],
    ["java", "is", "powerful"]
]

# Create and train Word2Vec model
model = Word2Vec(
    sentences,
    vector_size=10,
    window=2,
    min_count=1,
    workers=1,
    seed=42
)

# Find words similar to Java
print("\nWords similar to 'java':")
print(model.wv.most_similar("java", topn=3))

# Find words similar to Python
print("\nWords similar to 'python':")
print(model.wv.most_similar("python", topn=4))
```
